```python
import evaluate
import torch

```

```python
ds = [
    {
        "source": "Although the committee acknowledged the urgency of the situation, they postponed the final decision until additional data—most of which had not yet been verified—could be examined.",
        "prediction": "The committee knew the situation was urgent, but they delayed the decision until they checked more data.",
        "reference": "The committee agreed the situation was urgent, but they waited to decide until they reviewed more data."
    },
    {
        "source": "Because the instructions were written in highly technical language, even experienced professionals found the new software difficult to use during its initial release.",
        "prediction": "The instructions were very technical, so even experts had trouble using the new software at first.",
        "reference": "Since the instructions were complicated, the software was hard to use for many professionals."
    },
    {
        "source": "The researcher argued that the prevailing economic models, while elegant in theory, failed to account for the unpredictable behaviors observed in real markets.",
        "prediction": "The researcher said current economic models look good in theory but do not explain real market behavior.",
        "reference": "The researcher said that although the models look good on paper, they do not explain how people behave in real markets."
    }
]
```

```python
sari = evaluate.load("sari")

for sample in ds:
    score = sari.compute(
        predictions=[sample["prediction"]],
        references=[[sample["reference"]]],
        sources=[sample["source"]]
    )
    print(f"SARI score: {score['sari']}")
```
